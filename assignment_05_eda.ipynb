```python
# Variable 1: Market Segment
sns.countplot(data=df, x="market_segment")
plt.title("Market Segment")
plt.xlabel("Market Segment")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# Variable 2: Month of Arrival

sns.countplot(data=df, x="arrival_date_month", order=months)
plt.title("Month of Arrival")
plt.xlabel("Month of Arrival")
plt.ylabel("Number of Bookings")
months = ["January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"]
plt.xticks(rotation=45)
plt.show()
```

```python
# Variable 3: Repeat Guests vs. New Guests
plt.pie(df['is_repeated_guest'].value_counts(), labels=['New Guest', 'Repeated Guest'], autopct='%1.1f%%')
plt.title('Repeat vs. New Guests')
plt.show()
```

## 5. Bivariate Analysis

### Business framing:  

Stakeholders often ask: “What drives cancellations?” or “Do longer stays mean higher revenue?” Bivariate analysis helps you uncover those kinds of relationships.

### Do the following:
- Choose 2 relevant variable pairs (e.g., `lead_time` vs. `is_canceled`, or `adr` vs. `customer_type`)
- Use scatterplots, grouped bar plots, or boxplots to explore the relationships
- Interpret what these relationships could mean for the hotel business

```python
# Relationship 1: Lead Time vs. Average Daily Rate

sns.scatterplot(data=df, x='lead_time', y='adr')

plt.xlabel('Lead Time (Days)')
plt.ylabel('Average Daily Rate (ADR)')
plt.title('Lead Time vs. Average Daily Rate')
plt.show()
```

```python
# Relationship 2: Lead Time vs. Weeknight Stay Length
sns.scatterplot(data=df, x='lead_time', y='stays_in_week_nights')

plt.xlabel('Lead Time (Days)')
plt.ylabel('Weeknight Stay (Nights)')
plt.title('Lead Time vs. Weeknight Stay Length')
plt.show()
```
